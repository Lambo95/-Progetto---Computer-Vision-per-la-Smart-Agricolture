```python
import tensorflow as tf
import numpy as np

from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras.applications import ResNet50

from tensorflow.keras.utils import plot_model
```

```python
NUM_CLASSES = 5
IMG_HEIGHT = 224
IMG_WIDTH = 224
BATCH_SIZE = 16
```

```python
def create_dummy_data(samples=200):

    X = np.random.random(
        (samples, 224, 224, 3)
    ).astype(np.float32)

    y = np.random.randint(
        0,
        NUM_CLASSES,
        samples
    )

    return X, y
```

```python
X_train, y_train = create_dummy_data(200)
```

```python
base_model = ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)
```

```python
base_model.trainable = False
```

```python
model = models.Sequential([
    layers.Input(shape=(224,224,3)),
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(
        NUM_CLASSES,
        activation='softmax'
    )
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
plot_model(
    model,
    to_file="resnet50_agricolture.png",
    show_shapes=True,
    show_dtype=True,
    show_layer_names=True,
    expand_nested=True
)
```

```python
print("\nParametri Totali:")
print(model.count_params())

print("\nParametri Addestrabili:")
print(
    np.sum(
        [
            np.prod(v.shape)
            for v in model.trainable_weights
        ]
    )
)

print("\nParametri Congelati:")
print(
    np.sum(
        [
            np.prod(v.shape)
            for v in model.non_trainable_weights
        ]
    )
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=3,
    batch_size=16
)
```
